```python
import pandas as pd
import os

# โค้ดจะลองหาไฟล์ทั้งในกรณีที่อยู่โฟลเดอร์หลัก และโฟลเดอร์ notebooks
possible_paths = [
    "data/raw/csv/PTT.csv",       # กรณีที่รันอยู่โฟลเดอร์หลัก (SET50-Algo-Trading)
    "../data/raw/csv/PTT.csv"     # กรณีที่รันอยู่โฟลเดอร์ notebooks/
]

csv_path = None
for path in possible_paths:
    if os.path.exists(path):
        csv_path = path
        break

if csv_path:
    # โหลดไฟล์และตั้งให้คอลัมน์ date เป็น Index
    df_ptt = pd.read_csv(csv_path, index_col="date", parse_dates=True)
    display(df_ptt.tail())
else:
    print(f"หาไฟล์ไม่เจอ โฟลเดอร์ทำงานปัจจุบันของคุณคือ: {os.getcwd()}")
```
